# What is a frequency distribution? 
A frequency distribution represents the count (frequency) of values within a data set, the frequency of which represents the number of times a certain value appears within that data set. You can represent frequency distributions graphically with histograms, which can chart the frequncy of each discreet value independently, or via grouping into sets of a given range. 


# What is the mode of data? 
The mode represents whichever value in a given appears most frequently.

# Can we calculate the mean and the median for a nominal value? How about mode? 
We cannot calculate the mean of nominal values because mean calculations rely on quantitative information, which nominal values lack. We can, however, calculate the mode, which denotes the value that appears most frequently.
